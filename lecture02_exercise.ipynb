## Setup — Run this cell first


```python
import pandas as pd
import numpy as np

# World Happiness Report 2023 — representative data
# Source: https://www.kaggle.com/datasets/ajaypalsinghlo/world-happiness-report-2023

df = pd.read_csv('../data/world_happiness_2023.csv')
df.columns = ['Country','Region','Happiness_Score','GDP','Social_Support',
              'Life_Expectancy','Freedom','Generosity','Corruption']


print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
df.head()
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

## Task 1 — Regional Comparison Bar Chart

**What to build:** A horizontal bar chart showing the **average happiness score by region**, sorted from highest to lowest.

**Requirements:**
- Horizontal orientation (category names are long)
- Sorted by score, descending (so the happiest region is at the top)
- Zero baseline on x-axis
- At least one design choice that goes beyond the Plotly default (colour, annotation, labels, etc.)
- An insight title that answers: *which region stands out and why does it matter?*

**Hint:** Use `df.groupby('Region')['Happiness_Score'].mean()` to compute the averages.


```python
# Task 1: Regional comparison bar chart
# -------------------------------------

# Step 1: Compute average happiness score by region
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))  # sort for horizontal bar

print(region_avg)

# Step 2: Build your chart
# YOUR CODE HERE

# Highlight the region with the highest average happiness score
highest_region = region_avg['Happiness_Score'].idxmax()

colors = [
    '#167D8D' if i == highest_region else '#CFD8DC'
    for i in region_avg.index
]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=region_avg['Happiness_Score'],
    y=region_avg['Region'],
    orientation='h',
    marker_color=colors,
    text=region_avg['Happiness_Score'].round(2),
    textposition='outside',
    cliponaxis=False
))

# Insight title
top_region = region_avg.loc[highest_region, 'Region']
top_score = region_avg.loc[highest_region, 'Happiness_Score']

fig.update_layout(
    title=f'{top_region} has the highest average happiness score ({top_score:.2f})',
    xaxis_title='Average Happiness Score',
    yaxis_title='Region',
    xaxis=dict(
        range=[0, region_avg['Happiness_Score'].max() * 1.2],
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black',
        showgrid=True,
        gridcolor='#E8ECEF'
    ),
    yaxis=dict(
        categoryorder='array',
        categoryarray=region_avg.sort_values(
            'Happiness_Score'
        )['Region'].tolist()
    ),
    template='plotly_white',
    font=dict(family='Arial', size=13),
    showlegend=False,
    margin=dict(l=130, r=50, t=90, b=60)
)

fig.show()

```

## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build your chart
# YOUR CODE HERE


fig = go.Figure()

# Bottom 8 countries — muted orange
bottom_data = combined[combined['Group'] == 'Bottom 8']

fig.add_trace(go.Bar(
    x=bottom_data['Happiness_Score'],
    y=bottom_data['Country'],
    orientation='h',
    name='Bottom 8',
    marker_color='#D97757',
    text=bottom_data['Happiness_Score'].round(2),
    textposition='outside',
    cliponaxis=False
))

# Top 8 countries — teal
top_data = combined[combined['Group'] == 'Top 8']

fig.add_trace(go.Bar(
    x=top_data['Happiness_Score'],
    y=top_data['Country'],
    orientation='h',
    name='Top 8',
    marker_color='#167D8D',
    text=top_data['Happiness_Score'].round(2),
    textposition='outside',
    cliponaxis=False
))

# Global average reference line
fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='#333333',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

# Calculate the gap between the average scores of the two groups
top_avg = top8['Happiness_Score'].mean()
bottom_avg = bottom8['Happiness_Score'].mean()
gap = top_avg - bottom_avg

fig.update_layout(
    title=(
        f'The happiest 8 countries average {gap:.2f} points '
        'above the least happy 8'
    ),
    xaxis_title='Happiness Score',
    yaxis_title='Country',
    xaxis=dict(
        range=[0, combined['Happiness_Score'].max() * 1.25],
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black',
        showgrid=True,
        gridcolor='#E8ECEF'
    ),
    yaxis=dict(
        categoryorder='array',
        categoryarray=combined['Country'].tolist()
    ),
    template='plotly_white',
    font=dict(family='Arial', size=12),
    legend_title='Country Group',
    barmode='overlay',
    margin=dict(l=130, r=70, t=110, b=60)
)

fig.show()


```

## Done? Stretch Goal

If you finish both tasks with time to spare, try this:

**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
# Stretch goal — grouped bar chart
# YOUR CODE HERE




# Step 1: Select the five regions
selected_regions = [
    'Western Europe',
    'Latin America',
    'East Asia',
    'Sub-Saharan Africa',
    'South Asia'
]

# Step 2: Calculate average GDP and Freedom by region
region_factors = (
    df[df['Region'].isin(selected_regions)]
    .groupby('Region')[['GDP', 'Freedom']]
    .mean()
    .reindex(selected_regions)
    .reset_index()
)

print(region_factors)

# Step 3: Create grouped bars
fig3 = go.Figure()

fig3.add_trace(go.Bar(
    x=region_factors['Region'],
    y=region_factors['GDP'],
    name='GDP per capita',
    marker_color='#167D8D'
))

fig3.add_trace(go.Bar(
    x=region_factors['Region'],
    y=region_factors['Freedom'],
    name='Freedom',
    marker_color='#D97757'
))

# Step 4: Identify the region with the highest GDP average
highest_gdp_region = region_factors.loc[
    region_factors['GDP'].idxmax(), 'Region'
]

fig3.update_layout(
    title=(
        f'{highest_gdp_region} leads these regions in average GDP, '
        'while freedom scores provide a contrasting perspective'
    ),
    xaxis_title='Region',
    yaxis_title='Average factor score',
    barmode='group',
    template='plotly_white',
    font=dict(family='Arial', size=12),
    legend_title='Indicator',
    yaxis=dict(
        rangemode='tozero',
        showgrid=True,
        gridcolor='#E8ECEF',
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black'
    ),
    margin=dict(l=70, r=40, t=100, b=100)
)

fig3.show()
```
